# Image Operations using OpenCV

This notebook demonstrates fundamental image processing techniques using OpenCV, including color space transformations, filtering, thresholding, edge detection, morphological operations, watershed segmentation, and bitwise operations.

```python
import cv2
```

```python
img=cv2.imread(r"C:\Users\Hello\Downloads\images.jpg")
cv2.imshow('car',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Color Space Transformations

### Edge Detection

```python
import numpy as np
img3=cv2.imread(r"C:\Users\Hello\Downloads\fresh-green-apple-stockcake.jpg")
```

```python
blur = cv2.GaussianBlur(img3, (5, 5), 0)

# --- Add Canny Edge Detection ---
canny_edges = cv2.Canny(blur, 100, 200)

# Sobel in X direction
sobelx = cv2.Sobel(blur, cv2.CV_64F, 1, 0, ksize=3)

# Sobel in Y direction
sobely = cv2.Sobel(blur, cv2.CV_64F, 0, 1, ksize=3)

# Convert to absolute values
sobelx = cv2.convertScaleAbs(sobelx)
sobely = cv2.convertScaleAbs(sobely)

# Combine X and Y edges
sobel_combined = cv2.addWeighted(sobelx, 0.5, sobely, 0.5, 0)

# Display results

cv2.imshow('Original', img3)
cv2.imshow('Canny Edges', canny_edges)  # Now works!
cv2.imshow('Sobel X', sobelx)
cv2.imshow('Sobel Y', sobely)
cv2.imshow('Sobel Combined', sobel_combined)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Thresholding

```python
# Convert to grayscale
gray = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

# 1. Simple Thresholding (Unpack ret, thresh)

_, simple_thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 2. Adaptive Thresholding (Returns matrix directly)

adaptive_thresh = cv2.adaptiveThreshold(
    gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2
)

# 3. Otsu’s Thresholding (Unpack ret, thresh)

_, otsu_thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Display results

cv2.imshow('Original', img3)
cv2.imshow('Grayscale', gray)
cv2.imshow('Simple Threshold', simple_thresh)
cv2.imshow('Adaptive Threshold', adaptive_thresh)
cv2.imshow("Otsu's Threshold", otsu_thresh)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Morphological Operations

```python
img4=cv2.imread(r"C:\Users\Hello\Downloads\b3372d4bd8515bfeb7cfe8fc203fed74.jpg")
```

```python

gray = cv2.cvtColor(img4, cv2.COLOR_BGR2GRAY)

# --- FIX HERE: Unpack the tuple with `_, binary` ---
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Define kernel (structuring element)
kernel = np.ones((5, 5), np.uint8)

# 1. Erosion

erosion = cv2.erode(binary, kernel, iterations=1)

# 2. Dilation

dilation = cv2.dilate(binary, kernel, iterations=1)

# 3. Opening (Erosion + Dilation)

opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

# 4. Closing (Dilation + Erosion)

closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# Display results

cv2.imshow('Original', img4)
cv2.imshow('Binary', binary)
cv2.imshow('Erosion', erosion)
cv2.imshow('Dilation', dilation)
cv2.imshow('Opening', opening)
cv2.imshow('Closing', closing)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Contour Detection

```python
img5=cv2.imread(r"C:\Users\Hello\Downloads\2-rupees-coin-india-2011-DWN06W.jpg")
```

```python
gray = cv2.cvtColor(img5, cv2.COLOR_BGR2GRAY)

# Blur to reduce noise
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# --- FIX HERE: Unpack the tuple with `_, thresh` ---
_, thresh = cv2.threshold(blur, 127, 255, cv2.THRESH_BINARY)

# Find contours

contours, hierarchy = cv2.findContours(
    thresh,
    cv2.RETR_EXTERNAL,       # retrieve outer contours
    cv2.CHAIN_APPROX_SIMPLE  # compress contour points
)

# Draw contours & bounding boxes

output = img5.copy()
cv2.drawContours(output, contours, -1, (0, 255, 0), 2)
print("Number of contours detected:", len(contours))

for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    cv2.rectangle(output, (x, y), (x + w, y + h), (255, 0, 0), 2)

# Display results

cv2.imshow('Original', img5)
cv2.imshow('Threshold', thresh)
cv2.imshow('Contours & Bounding Boxes', output)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Image Transformations

```python

(h, w) = img3.shape[:2]


# 1. Translation (Shift Image)

tx, ty = 100, 50  # shift right and down
translation_matrix = np.float32([[1, 0, tx],
                                 [0, 1, ty]])

translated = cv2.warpAffine(img3, translation_matrix, (w, h))


# 2. Rotation

center = (w // 2, h // 2)
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)

rotated = cv2.warpAffine(img3, rotation_matrix, (w, h))


# 3. Affine Transformation

pts1 = np.float32([[50, 50], [200, 50], [50, 200]])
pts2 = np.float32([[10, 100], [200, 50], [100, 250]])

affine_matrix = cv2.getAffineTransform(pts1, pts2)
affine = cv2.warpAffine(img3, affine_matrix, (w, h))


# 4. Perspective Transformation


pts1 = np.float32([[100, 100], [400, 100], [100, 400], [400, 400]])
pts2 = np.float32([[0, 0], [300, 0], [0, 300], [300, 300]])

perspective_matrix = cv2.getPerspectiveTransform(pts1, pts2)
perspective = cv2.warpPerspective(img3, perspective_matrix, (300, 300))

# Display results

cv2.imshow('Original', img3)
cv2.imshow('Translated', translated)
cv2.imshow('Rotated', rotated)
cv2.imshow('Affine', affine)
cv2.imshow('Perspective (Document Scan)', perspective)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Feature Detection & Matching

```python
img6=cv2.imread(r"C:\Users\Hello\Downloads\images (6).jpg")
img7=cv2.imread(r"C:\Users\Hello\Downloads\images (5).jpg")
```

```python
gray1 = cv2.cvtColor(img6, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img7, cv2.COLOR_BGR2GRAY)

# 1. ORB (Fast & Free)

orb = cv2.ORB_create()

kp1_orb, des1_orb = orb.detectAndCompute(gray1, None)
kp2_orb, des2_orb = orb.detectAndCompute(gray2, None)

bf_orb = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches_orb = bf_orb.match(des1_orb, des2_orb)

matches_orb = sorted(matches_orb, key=lambda x: x.distance)

orb_result = cv2.drawMatches(
    img6, kp1_orb,
    img7, kp2_orb,
    matches_orb[:20], None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# 2. SIFT (Needs OpenCV contrib)

sift = cv2.SIFT_create()

kp1_sift, des1_sift = sift.detectAndCompute(gray1, None)
kp2_sift, des2_sift = sift.detectAndCompute(gray2, None)

bf_sift = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)
matches_sift = bf_sift.match(des1_sift, des2_sift)

matches_sift = sorted(matches_sift, key=lambda x: x.distance)

sift_result = cv2.drawMatches(
    img6, kp1_sift,
    img7, kp2_sift,
    matches_sift[:20], None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# 3. Display results

cv2.imshow('ORB Matches', orb_result)
cv2.imshow('SIFT Matches', sift_result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Image Segmentation

```python
img8=cv2.imread(r"C:\Users\Hello\Downloads\2-rupees-coin-india-2011-DWN06W.jpg")
```

```python

gray = cv2.cvtColor(img8, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)

_, thresh = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# Remove noise (opening)
kernel = np.ones((3, 3), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=2)

# Sure background area
sure_bg = cv2.dilate(opening, kernel, iterations=3)

# Distance Transform & Watershed Steps
dist_transform = cv2.distanceTransform(opening, cv2.DIST_L2, 5)
_, sure_fg = cv2.threshold(dist_transform, 0.7 * dist_transform.max(), 255, 0)

# Unknown region
sure_fg = np.uint8(sure_fg)
unknown = cv2.subtract(sure_bg, sure_fg)

# Marker labelling
_, markers = cv2.connectedComponents(sure_fg)
markers = markers + 1
markers[unknown == 255] = 0

# Apply Watershed
markers = cv2.watershed(img8, markers)
img8[markers == -1] = [0, 0, 255]  

# Display results

cv2.imshow('Watershed Result', img8)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## BitWise Operators

```python
img9=cv2.imread(r"C:\Users\Hello\Downloads\images (9).jpg")
img10=cv2.imread(r"C:\Users\Hello\Downloads\images (10).jpg")
```

```python
img10 = cv2.resize(img10, (img9.shape[1], img9.shape[0]))

# Create a mask (example: rectangle)
mask = np.zeros(img9.shape[:2], dtype="uint8")
cv2.rectangle(mask, (50, 50), (300, 300), 255, -1)

# Bitwise AND
bit_and = cv2.bitwise_and(img9, img9, mask=mask)

# Bitwise OR
bit_or = cv2.bitwise_or(img9, img10)

# Bitwise XOR
bit_xor = cv2.bitwise_xor(img9, img10)

# Bitwise NOT
bit_not = cv2.bitwise_not(img9)

# Show results
cv2.imshow("Original Image 9", img9)
cv2.imshow("Mask", mask)
cv2.imshow("Bitwise AND", bit_and)
cv2.imshow("Bitwise OR", bit_or)
cv2.imshow("Bitwise XOR", bit_xor)
cv2.imshow("Bitwise NOT", bit_not)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
